```python
from textblob import TextBlob
import pandas as pd
```

```python
reviews = [
    "This product is excellent",
    "I am very happy with this product",
    "This product is terrible",
    "I hate this product",
    "The product is okay"
]
```

```python
results = []
```

```python
for review in reviews:
    score = TextBlob(review).sentiment.polarity

    if score > 0:
        sentiment = "Positive"
    elif score < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    results.append({
        "Review": review,
        "Score": score,
        "Sentiment": sentiment
    })
```

```python
df = pd.DataFrame(results)
```

```python
df
```

```python
df["Sentiment"].value_counts()
```

```python
import matplotlib.pyplot as plt

df["Sentiment"].value_counts().plot(kind="bar")

plt.title("Customer Sentiment Analysis")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=0)
plt.show()
```

##Key Findings
1. Identification of Customer Sentiment:
   The sentiment analysis successfully classifies customer reviews into three categories: positive, negative, and neutral. This helps understand customers' opinions about a product or service.
2. Positive Customer Feedback:
   Reviews such as "This product is excellent," "I love this product," and "The service is very good" express positive opinions. These reviews indicate customer satisfaction and suggest that customers have a favorable impression of the product or service.
3. Negative Customer Feedback:
   Reviews such as "This product is terrible," "I hate this product," and "The service is bad" express negative opinions. These reviews can help businesses identify possible problems and areas that need improvement.
4. Neutral Customer Feedback:
   The review "The product is okay" expresses a relatively neutral opinion. However, the TextBlob sentiment tool may classify it as positive depending on its calculated polarity score. Therefore, the classification depends on the sentiment analysis model rather than human interpretation alone.
5. Sentiment Polarity Scores:
   The TextBlob library calculates a polarity score for each review. The score ranges from -1 to +1. A score below zero indicates negative sentiment, a score above zero indicates positive sentiment, and a score of zero indicates neutral sentiment.
6. Visual Representation of Results:
   A bar chart is created to display the number of positive, negative, and neutral reviews. This graphical representation makes it easier to compare the sentiment categories and understand the overall distribution of customer opinions.
7. Understanding Customer Satisfaction:
   The analysis provides an overview of customer feedback. A higher number of positive reviews may suggest greater customer satisfaction, while a higher number of negative reviews may indicate potential concerns that require further investigation.
8. Business Decision-Making:
   The results can help businesses understand customer preferences, improve product quality, enhance customer service, and develop better marketing strategies. However, the small sample of reviews used in this project is intended for demonstration and cannot represent all customers.
9. Use of Python Libraries:
   The project demonstrates how Python libraries such as TextBlob, Pandas, and Matplotlib can be used together to analyze text, organize results into a DataFrame, and visualize sentiment categories.
10. Limitations of Sentiment Analysis:
    Automated sentiment analysis may not always understand sarcasm, context, or complex expressions correctly. Therefore, the results should be treated as useful indicators of customer opinions rather than perfectly accurate judgments.
