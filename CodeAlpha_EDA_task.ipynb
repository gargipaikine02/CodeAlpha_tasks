```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Name": ["gargi", "gouri", "riya","trupti"],
    "Age": [19, 20, 19, 20],
    "Marks": [85, 90, 75, 95]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df["Marks"].mean()
```

```python
df.loc[df["Marks"].idxmax()]
```

```python
plt.bar(df["Name"], df["Marks"])
plt.title("Marks of Students")
plt.xlabel("Name")
plt.ylabel("Marks")
plt.show()
```

##Key Findings
1. Dataset Overview: The dataset contains information about five students, including their names, ages, and marks.
2. Data Structure: The dataset is organized into three columns: Name, Age, and Marks. Pandas helps display and understand the dataset structure and data types.
3. Statistical Analysis: The describe() function provides statistical information, including the average, minimum, maximum, and standard deviation of numerical columns.
4. Missing Values: The missing-value check shows that there are no missing values in the sample dataset, ensuring that all five student records are complete.
5. Average Marks: The average marks of the students are 85, indicating the overall performance of the students in the sample.
6. Highest Marks: Sneha secured the highest marks, scoring 95.
7. Data Visualization: A bar chart represents the marks of each student, making it easier to compare their performance and identify differences.

##Conclusion
Exploratory Data Analysis (EDA) is an important step in understanding and analyzing data. In this project, Python libraries such as Pandas and Matplotlib were used to examine a sample student dataset, summarize numerical information, check for missing values, and identify important results.
The analysis found that the average student score was 85, with Sneha achieving the highest score of 95. The dataset contained no missing values, making it suitable for this basic analysis. The bar chart provided a clear visual comparison of student marks.
Overall, this project demonstrates how EDA helps organize data, identify useful patterns, and summarize important information. These techniques can be applied to larger datasets to support data-driven decisions and further analysis.
